strings and lists can also be compared with these operators 
- if each character in a str matches another exactly then they are equal
- if each element in order in a list matches antoher exactly then they are equal

```python
list_1 = ["blue", "red", "green"]
list_2 = ["blue", "red", "green"]
list_3 = ["red", "green", "blue"]
list_4 = ["red", "green", "blue", "violet"]

print(list_1 == list_2) # true
print(list_1 == list_3) # false
print(list_3 == list_4) # false
print(list_3 != list_4) # true
```

`in` keyword allows us to test 
- whether a character appears in a str
- an object appears in a list
- a number appears in a numpy array

```python
keyword = "economic"
sentence = "The Federal Reserve makes forecasts about many economic outcomes"

print(keyword in sentence) #true

print("Federal" in sentence) #true
print("federal" in sentence) #false; case sensitive
```

```python
summer_months = ["June","July","August"]

print('June' in summer_months)
```
