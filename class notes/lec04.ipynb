```python
movie_rating = [7.5, 9.0, 4.5, 8.0, 6.5]

# stats
stats_mean = statistics.mean(movie_rating)
stats_median = statistics.median(movie_rating)

# numpy
numpy_mean = np.mean(movie_rating)
numpy_median = np.median(movie_rating)

print(f"stats_mean: {stats_mean:.2f}")
print(f"stats_median: {stats_median:.2f}")
print("-")
print(f"numpy_mean: {numpy_mean:.2f}")
print(f"numpy_median: {numpy_median:.2f}")
```
