compound data type = a type of data that combines multiple individual values or types into a single unit (e.h. lists, tuples, sets, numpy arrays)

## tuples

unlike lists, which are mutable, tuples are immutable and we cannot change them once they are created

```python
num_list = [5, 3, 2, 0] 

num_tuple = (5, 3, 2, 0) # can create without parenthesest too...

print(num_list)
print(num_tuple)
print()
print(type(num_list))
print(type(num_tuple))
```

```python
# concatenation works too

tuple_1 = (1, 2, 0, 9, 8)
tuple_2 = (-1, -50)

print(tuple_1 + tuple_2)
```

```python
# but this doesn't work bc (1) is an int
# we need a comma after the 1!

my_tuple = (1) + (2, 3, 4, 5, 6, 7, 8, 9)
print(my_tuple)
```

## sets

sets are an unordered collection of unique items
- created with curly braces {}
- mutable

```python
color_list = ["red", "blue", "red", "green", "blue", "red"]

color_set = {"red", "blue", "red", "green", "blue", "red"}

print(color_list)
print(color_set)
print()
print(type(color_list))
print(type(color_set))

# notice how duplicates are gone and the order isn't necessarily the same
```

```python
# also have set() to convert a list to a set. easy way to find unique items in a list

rolls = [3, 6, 1, 3, 3, 6, 2, 1]

unique_rolls = set(rolls)
print(rolls)
print(unique_rolls)
print(len(unique_rolls), "different numbers were rolled")
```

```python
# since theres no order, we can't index them like lists either 
print(color_set[0])

# but we can use `in` to check membership!
```

```python
color_set.add("purple")
print(color_set)

# adding an item that is already in the set does nothing
color_set.add("red")
print(color_set)

color_set.remove("blue")
print(color_set)
```

```python
print(type({}))
print(type(set()))
print(type({1}))
```

set operations: same operations in math
- **union** ($a \cup b$), written `a | b`: items in `a` **or** `b` (or both)
- **intersection** ($a \cap b$), written `a & b`: items in **both** `a` **and** `b`
- **difference** ($a \setminus b$), written `a - b`: items in `a` but **not** in `b`

**note:** use symbols `|` and `&`, not the words `or` and `and`
