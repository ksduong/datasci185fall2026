**Q2** In Lecture 6, we saw how to create a bar plot using `matplotlib.pyplot.bar`. We generated a bar plot showing the distribution of freshmen, sophomores, juniors, and seniors registered for DATASCI 151 this semester.

If we add the `color` argument, we can change the color of the bars, as shown in the code below.

```python
import matplotlib.pyplot as plt

categories = ["Freshman", "Sophomore", "Junior", "Senior"]

count = [22, 21, 31, 25]

plt.bar(x = categories, height = count, color = "hotpink")
plt.show()
```

In the cell below, a Boolean variable `is_pink` is defined. Using an if/else statement, write code that generates the bar plot with pink bars when `is_pink` is True. Otherwise, if it is `False` it should display a plot with the default color scheme, like in Lecture 6.

 Run the code with both `is_pink = False` and `is_pink = True` to make sure it works for both cases.

```python
categories = ["Freshman", "Sophomore", "Junior", "Senior"]

count = [22, 21, 31, 25]

is_pink = False

if is_pink:
    plt.bar(x = categories, height = count, color = "hotpink")
    plt.show()
else:
    plt.bar(x = categories, height = count)
    plt.show()
```

**Q4** The code cell below generates a random integer between 1 and 20 (both 1 and 20 are possible values).

- Test if the number satisfies $10 < n < 15$

```python
import random 

lo = 1
hi = 20

random_int = random.randint(lo, hi)
print(random_int)

if 1 <= random_int < 5:
    print("1<=x<5")
elif 5<=random_int<=10:
    print("5<=x<=10")
elif 10<random_int<15:
    print("10<x<15")
else:
    print("x<=15")
```
