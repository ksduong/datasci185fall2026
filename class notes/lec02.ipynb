use lists to store multiple objects
- always use square [] brackets
- elements separated by commas
- use len()
- start at [0]
- can have elements with different types, including existing variables and lists

```python
# assigning a list to a new var will just give the same list a new name
floats_list = [1.0, 2.0, 3.0, 4.0]

new_list = floats_list
print(new_list)

new_var = floats_list[2]
print(new_var)

print("---")

floats_list[2] = 0.0

print(floats_list)
print(new_list) # updated when floats_list updated
print(new_var)
```
