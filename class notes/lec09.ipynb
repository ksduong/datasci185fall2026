dictionaries are a compound data type indexed by keys instead of positions!

we use `key : value` pairs, with entries separated by commas

=> e.g. keys are the indices, values are the elements

```python
import pandas as pd
import random
```

```python
skill_scores = {
    "Abed" : 3.24, 
    "Britta" : 1.51, 
    "Jeff" : 0.51, 
    "Shirley" : 6.72, 
    "Annie" : 6.72, 
    "Troy" : 3.66, 
    "Pierce" : 0.58
    }

print(skill_scores["Jeff"])
```

```python
skill_scores["Jeff"] = 2.51

print(skill_scores)
```

```python
print(skill_scores.keys())
print(skill_scores.values())
```

we loop over a dictionary by looping over its keys

```python
for key in skill_scores.keys():
    name = key
    score = skill_scores[name]
    print(f"Lab skills for {name}:", score)
```

```python
# if you just try to loop over the dict itself, Python will loop over the keys automatically

for key in skill_scores:
    print(key)
```

- `.keys()` and `.values()` *look* like lists, but they are not (check the `type` below!)
- We can convert them into real lists with `list( )`
- This matters when we want to use list tools on them - for example `random.shuffle` from last class

```python
name_list = list(skill_scores.keys())

print(type(skill_scores.keys()))
print(type(name_list))
print(name_list)

# the same works for the values
score_list = list(skill_scores.values())
print(score_list)
```

```python
# since name_list is a list, we can shuffle it into a random order
random.shuffle(name_list)
print(name_list)
```

dictionaryception

```python
abed = {"month": "March", "day": 23, "year": 2007, "lab_skills": 3.24}

print(abed["month"])
print(abed["lab_skills"])
```

```python
study_group = {
    "Abed":   {"month": "March", "day": 23, "year": 2007, "lab_skills": 3.24},
    "Britta": {"month": "July",  "day": 14, "year": 1998, "lab_skills": 1.51},
}

print(study_group["Abed"])
```

to reach a single piece of information, we do a **two-step lookup**, reading left to right:
- `study_group["Abed"]` is itself a dictionary...
- ...so we can ask *that* dictionary for `["lab_skills"]`

Compare this with `birthdays[2][1]` from Lecture 6: same idea, but now with names instead of positions.

```python
print(study_group["Abed"]["lab_skills"])
print(study_group["Britta"]["month"])
```

making (small) dfs with dicts
- keys are columns
- for each key, define the value as a list whcih will contain the entries (lists have to be te same length)

```python
states = ["Wisconsin", "Georgia", "Delaware", "Illinois", "Oregon"]

# Wisconsin was 30th state admitted to the U.S., Georgia was 4th state admitted etc.
state_id = [30, 4, 1, 21, 33]

capitals = ["Madison", "Atlanta", "Dover", "Springfield", "Salem"]

# year admitted to the U.S.
admit_date = ["1848", "1788", "1787", "1818", "1859"]

# number of electoral votes for each state
electoral_votes = [10, 16, 3, 19, 8]

state_dict = {"state_id": state_id, "state_name" : states, "capital_city" : capitals, 
                "year_admitted" : admit_date, "electoral_college_votes" : electoral_votes}


df_states = pd.DataFrame(state_dict)

display(df_states)
```
