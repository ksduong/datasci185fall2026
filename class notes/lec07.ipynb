## for loops

```python
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

print("Dear customer, we are writing about your " + car_list[0] + " vehicle.")
print("Dear customer, we are writing about your " + car_list[1] + " vehicle.")
print("Dear customer, we are writing about your " + car_list[2] + " vehicle.")
print("Dear customer, we are writing about your " + car_list[3] + " vehicle.")

# Here we're concatenating strings with a "+" sign
```

```python
# with a for loop:

car_list = ["KIA", "Ferrari", "Ford", "Tesla"]

for car in car_list:
    print(f"Dear customer, we are writing about your {car} vehicle.")

# essentially loop goes through car = car_list[0], print(); car = car_list[1], print(); ... to car_list[-1]
```

sometimes it's good to know where we are in the list

```python
# method one: create an extra var to track
car_list = ["KIA", "Ferrari", "Ford", "Tesla"]
index = 0

for car in car_list:
    print(f"Dear {car} customer, your position is {index} on the waitlist.")
    index += 1

# we can add one to index if we want it to count like a human lol
```

```python
# method two: use range()

car_list = ["KIA", "Ferrari", "Ford", "Tesla"]
N = len(car_list)

for index in range(N):
    print(f"Dear {car_list[index]} customer, your position is {index + 1} on the waitlist.")

# essentially generates a sequence of integers the length of car_list from 0 to N-1. we use this range to iterate through the list by indexing car_list[index]
# range is an immutable sequence object that calculates the next number at each step of an iteration (lazy) rather than generating them all at once

# good to make index human counting INSIDE the forloop since we need the normal python list indexing for range
```

```python
print(N)
print(range(N))
print(type(range(N)))
range_list = list(range(N))
print(range_list)
```

## while loops

- iterate over a loop as long as some condition is true
- used if we don't know how many times to iterate beforehand



```python
# Specify a threshold level
import matplotlib.pyplot as plt

eps = 0.025

x_list = []
y_list = []


x = 1
while 1/x > eps:
    x_list.append(x)
    y_list.append(1/x)
    x = x + 1

plt.scatter(x = x_list, y = y_list)
plt.title("Graph of f(x) = 1/x")
plt.xlabel('x')
plt.ylabel("y")
plt.show()
```
