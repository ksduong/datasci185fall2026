We will build up to the full simulation one piece at a time.

**Q2** Now do that for every person in the room. Starting from an empty list, use a `for` loop to build a new list called `room`, where each entry is a copy of a real person but with a random month and day.

Print the first few entries and check that `len(room)` is 50.

*(We keep the `[name, month, day, year]` format so that later we can say exactly **who** shares a birthday.)*

```python
room = []
x = 0

for person in birthdays:
    copy = birthdays[x].copy()
    copy[1] = random.choice(months)
    copy[2] = random.choice(np.arange(1, 30).tolist())
    room.append(copy)
    x+=1

len(room)

print(room[0])
print(birthdays[0])

print(room[1])
print(birthdays[1])
```
