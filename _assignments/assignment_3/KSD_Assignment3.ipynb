**(b)** Imagine flipping a coin repeatedly until we get 10 heads (in total, not in a row). How many tails will you see before seeing the 10th head?

The **negative binomial distribution** describes exactly this. Given

- `n` = the number of heads needed to stop the experiment, and
- `p` = the probability of heads on a single flip,

it tells us the probability of getting each possible number of tails before the experiment ends (0 tails, 1 tail, 2 tails, ... ).

The probability of exactly `k` tails before seeing `n` heads is

$$P(k \text{ tails}) = \frac{(k+n-1)!}{k!\,(n-1)!}\, p^{k}\,(1-p)^{n}$$

Use the scipy function `scipy.stats.nbinom` to plot the probability mass function (pmf) over the range $k=0, 1, 2, ..., 9, 10$ when we stop after flipping $n=10$ heads of a fair coin (so $p = 0.5$).

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

n = 10
p = 0.5
k = np.arange(0, 11)

probability = stats.nbinom.pmf(k, n, p)

print(probability)

plt.bar(k, probability)
plt.xlabel("Number of Tails")
plt.ylabel("Probability")
plt.title("Negative Binomial PMF")
plt.show()
```

**(c)** The binomial and negative binomial distributions are **discrete** random variables - they take on integer values only. We can visualize these with the probability mass function (pmf). **Continuous** random variables can take any value within an interval of the real numbers. To visualize these distributions, we use the probability *density* function (pdf).

`scipy.stats` also supports continuous random variables, such as the normal distribution. We can generate a sequence of points corresponding to possible values of the distribution, evaluate the pdf at those points, and plot the result.

Do the following:

- Use `np.linspace` to generate an array named `x` containing 100 evenly spaced points from $-5$ to $5$.

- Pass `x` into the function `stats.norm.pdf` to evaluate the probability density function (pdf) of the standard normal distribution.

- Use `plt.plot` to create a plot of the standard normal pdf over the range $[-5, 5]$. Add labels and a title.

- Pass `x` again into the function `stats.norm.pdf` but add the keyword arguments `loc = 1`, `scale = 2` to evaluate the pdf of a normal distribution with mean = 1 and standard deviation = 2.

- Use `plt.plot` to create a plot of the pdf of a normal random variable with mean = 1 and standard deviation = 2 over the range $[-5, 5]$. The plot should be generated separately from the first. Add labels and a title as well.

```python
x = np.linspace(-5, 5, 100)

prob1 = stats.norm.pdf(x)

plt.plot(x, prob1)
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Standard Normal Distribution")
plt.show()

prob2 = stats.norm.pdf(x, loc=1, scale=2)

plt.plot(x, prob2)
plt.xlabel("x")
plt.ylabel("Probability Density")
plt.title("Shifted Normal Distribution")
plt.show()
```

**(d)** The code below randomly selects two names from the list. Using comparison operators and conditional statements, write a block of code that:

- prints `f"{name1} comes before {name2}"` if `name1` comes alphabetically before `name2`

- prints `f"{name2} comes before {name1}"` if `name2` comes alphabetically before `name1`

- prints `"The names are the same"` if `name1` is the same as `name2`.

```python
import random

string_list = ["Frankie", "Nathan", "Peter", "Abed", "Britta", "Elroy"]
name1 = random.sample(string_list, 1)
name2 = random.sample(string_list, 1)

if name1 < name2:
    print(f"{name1[0]} comes before {name2[0]}")
elif name2 < name1:
    print(f"{name2[0]} comes before {name1[0]}")
else:
    print("The names are the same")
```
